```python
from penv.ast import *
import penv.evaluator as ev

rho = ev.empty_env()

term = App(
    Ext(NameConst("value-of-x"), "x", Id()),
    NameConst("x")
)

ev.eval(term, rho)
```

```python
term = Comp(
    Var("x"),
    Ext(NameConst("M"), "x", Id())
)

ev.eval(term, rho)
```

```python
term = App(
    App(
        Lam("x", Lam("y", Id())),
        NameConst("M")
    ),
    NameConst("N")
)

env_result = ev.eval(term, rho)
env_result
```

```python
ev.lookup(env_result, "x")
```

```python
env_mn = App(
    App(
        Lam("x", Lam("y", Id())),
        Lam("z", Var("z"))
    ),
    NameConst("N")
)

term = Comp(
    App(Var("x"), Var("y")),
    env_mn
)

ev.eval(term, rho)
```

```python
term = Comp(
    Var("x"),
    Lam(
        "i",
        If(
            Equal(Var("i"), NameConst("x")),
            NameConst("M"),
            NameConst("N")
        )
    )
)

ev.eval(term, rho)
```
